### Задача 1. Первый взгляд
Загрузите `orders.csv` в переменную `df`.

```python
# Задача 1
df = pd.read_csv("orders.csv")

print("Форма:", df.shape)

n_karaganda = (df["city"] == "Караганда").sum()
print("Заказов из Караганды:", n_karaganda)

acc_products = df[df["category"] == "Аксессуары"]["product"].unique()
print(acc_products)
print("Разных аксессуаров:", len(acc_products))

# Вывод: в таблице 500 заказов и 9 столбцов; из Караганды 72 заказа, а аксессуаров в ассортименте всего 4 вида - клавиатура, наушники, power bank и мышь.
```

### Задача 2. Дорогие заказы Шымкента
Добавьте столбец `revenue` = `price` × `qty`. Выведите 5 самых дорогих **оплаченных** заказов из Шымкента: только столбцы `date`, `product`, `qty`, `revenue`.

```python
# Задача 2
df["revenue"] = df["price"] * df["qty"]
is_paid = df["status"] == "оплачен"

shym = df[is_paid & (df["city"] == "Шымкент")]
top5 = shym.sort_values("revenue", ascending=False).head(5)
print(top5[["date", "product", "qty", "revenue"]])

# Вывод: самые дорогие заказы Шымкента — ноутбуки: топ-1 — два ASUS VivoBook 15 на 638 000 ₸, ещё три раза брали MacBook Air M3 по 549 000 ₸.
```

### Задача 3. Покупают помногу?
1. Сколько заказов, в которых купили **2 штуки или больше**? Какая это доля всех заказов в %?
2. Какой товар чаще всего покупают сразу **по 3 штуки**?

```python
# Задача 3
many = df["qty"] >= 2
print("Заказов с 2+ штуками:", many.sum())
print(f"Доля: {many.mean() * 100:.1f}%")

three = df[df["qty"] == 3]["product"].value_counts()
print(three.head())
print("Чаще всего по 3 штуки берут:", three.idxmax())

# Вывод: 116 заказов (23.2%) — покупка 2+ штук; по 3 чаще всего берут наушники JBL Tune 520 (13 раз) — дешёвые аксессуары берут «на всех».
```
